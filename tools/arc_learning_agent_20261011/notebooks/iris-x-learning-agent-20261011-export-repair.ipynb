This successor prunes the fixed installed payload subtree before descent and retains only bounded hash-bound environment proof files. Other output symlinks, changed files and missing proofs remain rejected.

This is an inactive source repair, code Dataset v4. MANUAL_START and ENABLE_MODEL default to False. The original v6 ERROR, original predictions and scoring products remain immutable. CPU-only recovery separately packages those existing products without regenerating, reranking or rescoring. No new GPU run is authorized by this release.

The prior model, training, controller, scorer, all other seven Dataset pins, Docker image, ARC3 association, resource limits and pre-loader supervisor remain unchanged. A future active run requires separate explicit authorization.

```python
"""Default-inert, stdlib-only deadline supervisor for the whole reviewed notebook.

The notebook coordinator never loads model weights. One owned child performs
source admission, offline installation and the complete session. Immutable
process-guard source is reused, including pidfd/subreaper/parent-death handling.
"""
from __future__ import annotations
import hashlib
import importlib.util
import json
import math
import os
from pathlib import Path
import sys
import time

MAX_SESSION_SECONDS = 43200
CLEANUP_RESERVE_SECONDS = 30


def _need(value, message):
    if not value: raise ValueError(message)


def _canonical(value):
    return (json.dumps(value, sort_keys=True, separators=(',', ':'), ensure_ascii=False, allow_nan=False)+'\n').encode()


def _hash(raw): return hashlib.sha256(raw).hexdigest()


def _path(value):
    _need(type(value) is str and value and len(value)<=4096 and not any(ord(c)<32 for c in value), 'invalid path')
    _need('\\' not in value and ':' not in value and '..' not in value.split('/'), 'ambiguous path')
    return Path(os.path.abspath(value))


def _overlap(a,b): return a==b or a in b.parents or b in a.parents


def _links(path):
    _need(not any(p.is_symlink() for p in (path,*path.parents)), 'symlink path')


def _fresh_write(path,raw):
    fd=os.open(path,os.O_WRONLY|os.O_CREAT|os.O_EXCL|os.O_NOFOLLOW,0o600)
    with os.fdopen(fd,'wb') as stream:
        stream.write(raw);stream.flush();os.fsync(stream.fileno())


def _load_guard(path,expected_sha):
    _need(_hash(path.read_bytes())==expected_sha,'guard bytes changed')
    spec=importlib.util.spec_from_file_location('_irisx_owned_notebook_guard',path)
    module=importlib.util.module_from_spec(spec);spec.loader.exec_module(module)
    return module


def _run_owned(argv, *, guard_module, launcher_root, output_root, started_epoch,
               hard_deadline_epoch, cleanup_reserve=CLEANUP_RESERVE_SECONDS, poll_seconds=0.2):
    """Only called with the fixed reviewed bootstrap, or explicit CPU tests."""
    remaining=hard_deadline_epoch-time.time()
    _need(remaining>cleanup_reserve, 'deadline reserve already reached before child launch')
    hard_mono=time.monotonic()+remaining
    work_mono=hard_mono-cleanup_reserve
    forced=False;reason=None;next_progress=0.0
    with (launcher_root/'stdout.txt').open('xb') as out,(launcher_root/'stderr.txt').open('xb') as err:
        guard=guard_module.OwnedProcessGuard.spawn(argv,evidence_dir=launcher_root/'owned-tree',
            stdout=out,stderr=err,cwd=str(Path.cwd()),close_fds=True,start_new_session=True,
            env={**os.environ,'PYTHONDONTWRITEBYTECODE':'1','HF_HUB_OFFLINE':'1','TRANSFORMERS_OFFLINE':'1'})
        try:
            while guard.poll() is None:
                if time.monotonic()>=work_mono or time.time()>=hard_deadline_epoch-cleanup_reserve:
                    forced=True;reason='absolute-session-deadline';guard.stop(timeout=min(10.0,max(0.1,hard_mono-time.monotonic())));break
                if time.monotonic()>=next_progress:
                    # Operational labels only. Never open scorer metrics or task answers.
                    marker=output_root/'worker/active-stage.json';stage='source-admission-or-bootstrap'
                    if marker.is_file() and not marker.is_symlink() and marker.stat().st_size<=65536:
                        state=json.loads(marker.read_bytes());stage=str(state.get('stage','unknown'))[:80]
                    print(json.dumps({'event':'LEARNING_PROGRESS','stage':stage,
                        'elapsed_seconds':round(max(0,time.time()-started_epoch),1),
                        'remaining_seconds':round(max(0,min(hard_mono-time.monotonic(),hard_deadline_epoch-time.time())),1)}),flush=True)
                    next_progress=time.monotonic()+30
                time.sleep(poll_seconds)
        except BaseException:
            forced=True;reason='coordinator-error';guard.stop(timeout=10);raise
        finally:
            proof=guard.closed_proof()
            if not proof.get('closed') or proof.get('remaining'):
                forced=True;reason=reason or 'unclosed-owned-tree';proof=guard.stop(timeout=10)
            completed_epoch=time.time()
            receipt={'format':'irisx/learning-notebook-closure-v1','started_epoch':started_epoch,
                'hard_deadline_epoch':hard_deadline_epoch,'completed_epoch':completed_epoch,
                'within_absolute_deadline':completed_epoch<=hard_deadline_epoch and time.monotonic()<=hard_mono,
                'forced':forced,'reason':reason,'process_tree_proof':proof,
                'scope':'owned notebook child execution; no VM/billing/queue SLA claim'}
            _fresh_write(launcher_root/'launcher-closure.json',_canonical(receipt))
    guard_module.verify_closed_proof(proof)
    if forced or not receipt['within_absolute_deadline'] or proof['returncode']!=0 or proof['worker_returncode']!=0 or proof['forced']:
        raise RuntimeError('owned notebook execution stopped or failed; inspect launcher closure/stdout/stderr')
    return receipt


def launch_reviewed_notebook(*, pins, config, loader_source, guard_source, expected_loader_sha256, expected_guard_sha256):
    """No task/model or source asset is read before the owned supervisor starts."""
    required={'manual_start','enable_model','profile','session_seconds','predispatch_epoch','output_root',
              'source_parent','input_root','code_root','data_root','training_root','model_config','seed'}
    _need(type(config) is dict and set(config)==required,'exact launcher config required')
    _need(type(config['manual_start']) is bool and type(config['enable_model']) is bool,'exact switches')
    if not config['manual_start']:return {'status':'DISABLED'}
    _need(config['profile']=='LEARNING_AGENT12H' and type(config['session_seconds']) is int and config['session_seconds']==MAX_SESSION_SECONDS,'frozen session cap')
    epoch=config['predispatch_epoch']
    if not config['enable_model'] and epoch is None:epoch=time.time()  # explicit CPU preflight only
    _need(type(epoch) in (int,float) and math.isfinite(epoch) and 0<=epoch<=time.time()+2,'active submission requires a finite predispatch epoch')
    hard_epoch=epoch+MAX_SESSION_SECONDS
    _need(hard_epoch-time.time()>60,'submission clock expired; no new session')
    _need(_hash(loader_source.encode())==expected_loader_sha256 and _hash(guard_source.encode())==expected_guard_sha256,'embedded executable source differs')
    model_config={} if config['model_config'] is None else config['model_config']
    _need(type(model_config) is dict and set(model_config)<={'input_root','explicit_roots'},'bounded bootstrap config')
    roots=model_config.get('explicit_roots',{});_need(type(roots) is dict,'explicit roots must be mapping')
    inputs=[_path(config['input_root']),_path(model_config.get('input_root',config['input_root']))]
    inputs += [_path(v) for v in [config['code_root'],config['data_root'],config['training_root'],*roots.values()] if v is not None]
    output=_path(config['output_root']);source=_path(config['source_parent'])
    launcher=output.parent/(output.name+'-launcher');environment=output.parent/(output.name+'-environment')
    writes=[output,source,launcher,environment]
    for i,a in enumerate(writes):
        for b in inputs+writes[i+1:]:_need(not _overlap(a,b),'launcher/source/output overlaps immutable input or another write root')
    for p in writes:_links(p)
    _need(not launcher.exists() and not output.exists(),'fresh launcher/output identity required')
    launcher.mkdir(parents=True,mode=0o700,exist_ok=False)
    guard_path=launcher/'owned_process_guard.py';_fresh_write(guard_path,guard_source.encode())
    driver=loader_source+'''\n\nif __name__ == '__main__':
    import hashlib,json,sys
    from pathlib import Path
    request_path,request_sha=sys.argv[1:]
    raw=Path(request_path).read_bytes()
    if hashlib.sha256(raw).hexdigest()!=request_sha:raise ValueError('launcher request changed')
    request=json.loads(raw);c=request['config']
    ready=load_reviewed_learning_source(pins=request['pins'],input_root=c['input_root'],
        code_root=c['code_root'],data_root=c['data_root'],training_root=c['training_root'],
        source_parent=c['source_parent'],output_root=c['output_root'],model_config={} if c['model_config'] is None else c['model_config'],resume=False)
    from tools.arc_learning_agent_20261011.session import run_session
    result=run_session(ready['output_root'],solver_root=ready['solver_root'],evaluator_root=ready['evaluator_root'],
        solver_manifest_sha256=ready['solver_manifest_sha256'],evaluator_manifest_sha256=ready['evaluator_manifest_sha256'],
        training_root=ready['training_root'],enable_model=c['enable_model'],bootstrap_config=c['model_config'],
        started_epoch=request['started_epoch'],config={'session_seconds':c['session_seconds'],'seed':c['seed'],'predispatch_clock':c['predispatch_epoch'] is not None})
    target=Path(request['launcher_root'])/'result.json'
    with target.open('x') as f:json.dump(result,f,sort_keys=True,allow_nan=False);f.flush();__import__('os').fsync(f.fileno())
    print(json.dumps({'status':result['status'],'run_id':result['run_id'],'score_status':result.get('score_status')}),flush=True)
    raise SystemExit(0 if result['status'] in ('COMPLETE','PARTIAL_SCORED','PREFLIGHT_ONLY') else 2)
'''
    driver_path=launcher/'bootstrap.py';_fresh_write(driver_path,driver.encode())
    request={'pins':pins,'config':config,'started_epoch':epoch,'hard_deadline_epoch':hard_epoch,'launcher_root':str(launcher),
             'loader_sha256':expected_loader_sha256,'guard_sha256':expected_guard_sha256,'driver_sha256':_hash(driver.encode())}
    raw=_canonical(request);request_path=launcher/'request.json';_fresh_write(request_path,raw)
    module=_load_guard(guard_path,expected_guard_sha256)
    receipt=_run_owned([sys.executable,'-I','-B',str(driver_path),str(request_path),_hash(raw)],guard_module=module,
        launcher_root=launcher,output_root=output,started_epoch=epoch,hard_deadline_epoch=hard_epoch)
    result_path=launcher/'result.json';_need(result_path.is_file() and result_path.stat().st_size<=1024*1024,'bounded child result required')
    result=json.loads(result_path.read_bytes())
    return {**result,'notebook_owner_closure':{'path':str(launcher/'launcher-closure.json'),'sha256':_hash(_canonical(receipt))}}

LEARNING_PINS = {'code': {'reference': 'danielwuu/iris-x-learning-agent-code-20261011/4', 'manifest_sha256': 'd6784cce76cc830488b3b2d29559309cd0b070df48639a34d5069d7b260ddf3c', 'archive_sha256': '894430037e6e2f41cff726f4ebcb4bcebde9e1df67198109e89c4cc34ddb3c6c'}, 'data': {'reference': 'danielwuu/iris-x-public120-data-20261010/1', 'manifest_sha256': 'c55073f8e9f44682dd7d7f31ee7f04f39754a9922ec246073555026452f047e3', 'role_manifests': {'evaluator': 'c7e8ad35770a66a71a22cccbbd0455c1c3bcc1a505dfb2e4c29819eec6209e33', 'solver': '549019b4b5cb47176f3abd3fcd9ca516e91b7d6e581dfc414a5a6452ff7b1f7f'}}, 'training': {'reference': 'danielwuu/iris-x-learning-training-20261011/1', 'manifest_sha256': 'a4b9da04d8690c57a3d0855553989fa055c5b9283ffc1784fe99413c8d1cb9d5'}}
LOADER_SOURCE = '"""Stdlib-only notebook loader. Nothing is read or written until explicitly called."""\nfrom __future__ import annotations\n\ndef load_reviewed_learning_source(*, pins, input_root, code_root, data_root, training_root, source_parent, output_root, model_config, resume=False):\n    import hashlib, json, os, stat, sys, zipfile\n    from pathlib import Path, PurePosixPath\n    def need(value,message):\n        if not value:raise ValueError(message)\n    def path(value,absolute=False):\n        need(type(value) is str and value and len(value)<=4096 and not any(ord(c)<32 for c in value),\'invalid path\')\n        need(\'\\\\\' not in value and \':\' not in value and \'..\' not in value.split(\'/\'),\'ambiguous/traversing path\')\n        need(not absolute or value.startswith(\'/\'),\'absolute input root required\')\n        return Path(\'/\'+os.path.abspath(value).lstrip(\'/\'))\n    def overlap(a,b):return a==b or a in b.parents or b in a.parents\n    def links(p):\n        for q in (*reversed(p.parents),p):need(not q.is_symlink(),\'symlink path forbidden\')\n    def read(p,limit):\n        links(p);need(stat.S_ISREG(p.stat().st_mode) and p.stat().st_size<=limit,\'bounded regular file required\')\n        return p.read_bytes()\n    def sha(raw):return hashlib.sha256(raw).hexdigest()\n    def pairs(rows):\n        result={}\n        for k,v in rows:\n            need(k not in result,\'duplicate JSON key\');result[k]=v\n        return result\n    def parse(raw):return json.loads(raw,object_pairs_hook=pairs,parse_constant=lambda v:(_ for _ in ()).throw(ValueError(\'nonfinite JSON\')))\n    need(type(resume) is bool,\'exact Boolean resume required\')\n    need(type(model_config) is dict and set(model_config)<= {\'input_root\',\'explicit_roots\'},\'invalid model config\')\n    explicit=model_config.get(\'explicit_roots\',{})\n    need(type(explicit) is dict and set(explicit)<= {\'runtime\',\'model\',\'wheels\',\'interpreter\',\'adapter\'},\'unknown model root\')\n    inputs=[path(input_root,True),path(model_config.get(\'input_root\',input_root),True)]\n    inputs += [path(v,True) for v in explicit.values() if v is not None]\n    for value in (code_root,data_root,training_root):\n        if value is not None:inputs.append(path(value,True))\n    source_parent,path_output=path(source_parent),path(output_root)\n    environment=path_output.parent/(path_output.name+\'-environment\')\n    writes=[source_parent,path_output,environment]\n    # Finish all lexical comparisons before any filesystem inspection or write.\n    for target in writes:\n        for immutable in inputs:need(not overlap(target,immutable),\'write path overlaps immutable input\')\n    for i,a in enumerate(writes):\n        for b in writes[i+1:]:need(not overlap(a,b),\'source/output/environment overlap\')\n    for target in writes:links(target)\n    def discover(reference,explicit_root):\n        if explicit_root is not None:return path(explicit_root,True)\n        owner,slug,version=reference.split(\'/\')\n        candidates=[inputs[0]/slug,inputs[0]/\'datasets\'/owner/slug]\n        present=[p for p in candidates if p.is_dir()]\n        need(len(present)==1,\'exactly one pinned Dataset mount required: \'+slug)\n        return present[0]\n    code=discover(pins[\'code\'][\'reference\'],code_root);data=discover(pins[\'data\'][\'reference\'],data_root)\n    training=discover(pins[\'training\'][\'reference\'],training_root)\n    links(code);links(data)\n    code_raw=read(code/\'manifest.json\',1024*1024)\n    need(sha(code_raw)==pins[\'code\'][\'manifest_sha256\'],\'code manifest hash differs\')\n    manifest=parse(code_raw)\n    need(manifest[\'format\']==\'irisx/learning-source-v1\' and manifest[\'reference\']==pins[\'code\'][\'reference\'],\'code reference differs\')\n    archive=manifest[\'archive\'];need(archive[\'name\']==\'source.zip\',\'unexpected archive\')\n    need(archive[\'sha256\']==pins[\'code\'][\'archive_sha256\'],\'notebook source archive differs\')\n    data_raw=read(data/\'public-data-material.json\',1024*1024)\n    need(sha(data_raw)==pins[\'data\'][\'manifest_sha256\'],\'data material manifest differs\')\n    dm=parse(data_raw);need(dm[\'reference\']==pins[\'data\'][\'reference\'] and dm[\'format\']==\'irisx/public120-data-material-v1\',\'data reference differs\')\n    need(dm[\'role_manifests\']==pins[\'data\'][\'role_manifests\'],\'data role identities differ\')\n    # Training source has an independent exact closed inventory; no evaluator file is read.\n    training_raw=read(training/\'manifest.json\',2*1024**2)\n    need(sha(training_raw)==pins[\'training\'][\'manifest_sha256\'],\'training manifest differs\')\n    tm=parse(training_raw)\n    need(tm[\'format\']==\'irisx/learning-training-material-v1\' and tm[\'reference\']==pins[\'training\'][\'reference\'],\'training source identity differs\')\n    need(tm[\'tasks\']==1000 and tm[\'public_evaluation_files\']==0,\'clean1000 training profile required\')\n    need(len(tm[\'files\'])<=1100 and sum(r[\'bytes\'] for r in tm[\'files\'].values())<=32*1024**2,\'training material cap\')\n    observed_training=set()\n    for member in training.rglob(\'*\'):\n        links(member)\n        if member.is_file():observed_training.add(member.relative_to(training).as_posix())\n        else:need(member.is_dir(),\'nonregular training member\')\n    need(observed_training==set(tm[\'files\'])|{\'manifest.json\'},\'training inventory differs\')\n    for name,row in tm[\'files\'].items():\n        pure=PurePosixPath(name)\n        need(not pure.is_absolute() and all(x not in (\'\',\'.\',\'..\') for x in name.split(\'/\')) and \'\\\\\' not in name,\'unsafe training member\')\n        body=read(training/name,2*1024**2)\n        need(len(body)==row[\'bytes\'] and sha(body)==row[\'sha256\'],\'training byte identity differs\')\n    # No evaluator member is opened here; evaluator validates it after sealing.\n    files=manifest[\'files\'];need(type(files) is dict and 1<=len(files)<=512,\'source inventory ceiling\')\n    need(set(manifest[\'source_identity\'])==set(files),\'closed source identity required\')\n    for name,row in files.items():\n        p=PurePosixPath(name)\n        need(type(name) is str and not p.is_absolute() and \'\\\\\' not in name and \':\' not in name and all(x not in (\'\',\'.\',\'..\') for x in name.split(\'/\')),\'unsafe source member\')\n        need(type(row[\'bytes\']) is int and 0<=row[\'bytes\']<=2*1024**2,\'source member size\')\n        need(manifest[\'source_identity\'][name]==row[\'sha256\'],\'source identity mismatch\')\n    need(sum(r[\'bytes\'] for r in files.values())<=16*1024**2,\'source expansion ceiling\')\n    content={}; allowed_metadata={\'manifest.json\',\'NOTICE.txt\'}\n    observed=set()\n    for member in code.rglob(\'*\'):\n        links(member)\n        if member.is_file():observed.add(member.relative_to(code).as_posix())\n        else:need(member.is_dir(),\'nonregular Dataset member\')\n    archive_path=code/\'source.zip\';links(archive_path)\n    if archive_path.is_file():\n        need(observed-allowed_metadata=={\'source.zip\'},\'ambiguous archive and extracted source inventory\')\n        raw=read(archive_path,16*1024**2)\n        need(sha(raw)==archive[\'sha256\'] and len(raw)==archive[\'bytes\'],\'source archive integrity\')\n        import io\n        with zipfile.ZipFile(io.BytesIO(raw)) as z:\n            members=z.infolist();need(len(members)==len(files) and len({m.filename for m in members})==len(members),\'duplicate/extra archive members\')\n            need(sum(m.file_size for m in members)<=16*1024**2,\'source expansion ceiling\')\n            for m in members:\n                name=m.filename\n                need(name in files,\'unsafe archive member\')\n                need(stat.S_IFMT(m.external_attr>>16) in (0,stat.S_IFREG) and not m.is_dir() and m.file_size<=2*1024**2,\'nonregular/large source member\')\n                value=z.read(m);need(len(value)==files[name][\'bytes\'] and sha(value)==files[name][\'sha256\'],\'source file integrity\')\n                content[name]=value\n        transport_mode=\'verified-archive\'\n    else:\n        # Kaggle may extract an uploaded ZIP. Accept only these exact two\n        # layouts, never recursive filename discovery or incomplete fallback.\n        layouts=[(\'\',code),(\'source/\',code/\'source\')]\n        matched=[(prefix,base) for prefix,base in layouts if observed-allowed_metadata=={prefix+n for n in files}]\n        need(len(matched)==1,\'exactly one closed extracted source inventory required\')\n        prefix,base=matched[0]\n        for name,row in files.items():\n            value=read(base/name,2*1024**2)\n            need(len(value)==row[\'bytes\'] and sha(value)==row[\'sha256\'],\'extracted source file integrity\')\n            content[name]=value\n        transport_mode=\'verified-extracted-source\' + (\'-directory\' if prefix else \'-root\')\n    source=source_parent/archive[\'sha256\'];links(source)\n    if source.exists():\n        actual={p.relative_to(source).as_posix() for p in source.rglob(\'*\') if p.is_file() or p.is_symlink()}\n        need(actual==set(content),\'existing source inventory differs\')\n        for name,value in content.items():need(read(source/name,2*1024**2)==value,\'existing source bytes differ\')\n    else:\n        source.mkdir(parents=True,exist_ok=False)\n        for name,value in content.items():\n            target=source/name;target.parent.mkdir(parents=True,exist_ok=True)\n            with target.open(\'xb\') as f:f.write(value);f.flush();os.fsync(f.fileno())\n    for key,module in tuple(sys.modules.items()):\n        if key==\'tools\' or key.startswith(\'tools.\') or key==\'irisx\' or key.startswith(\'irisx.\'):\n            origin=getattr(module,\'__file__\',None)\n            need(origin is not None and source in path(str(origin)).parents,\'conflicting imported project source; restart kernel\')\n    sys.dont_write_bytecode=True\n    sys.path[:0]=[str(source),str(source/\'src\')]\n    entry_provenance={\'format\':\'irisx/public120-notebook-source-admission-v1\',\n        \'code_reference\':pins[\'code\'][\'reference\'],\'code_manifest_sha256\':pins[\'code\'][\'manifest_sha256\'],\n        \'source_archive_sha256\':archive[\'sha256\'],\n        \'source_identity_sha256\':sha(json.dumps(manifest[\'source_identity\'],sort_keys=True,separators=(\',\',\':\'),ensure_ascii=True).encode()),\n        \'source_transport_mode\':transport_mode,\'source_archive_bytes_verified\':transport_mode==\'verified-archive\',\n        \'data_reference\':pins[\'data\'][\'reference\'],\'data_material_manifest_sha256\':pins[\'data\'][\'manifest_sha256\'],\n        \'solver_manifest_sha256\':pins[\'data\'][\'role_manifests\'][\'solver\'],\n        \'evaluator_manifest_sha256\':pins[\'data\'][\'role_manifests\'][\'evaluator\']}\n    return {\'training_root\':str(training/\'data/training\'),\'training_manifest_sha256\':pins[\'training\'][\'manifest_sha256\'],\n            \'source_root\':str(source),\'solver_root\':str(data/\'solver\'),\'evaluator_root\':str(data/\'evaluator\'),\n            \'solver_manifest_sha256\':pins[\'data\'][\'role_manifests\'][\'solver\'],\'evaluator_manifest_sha256\':pins[\'data\'][\'role_manifests\'][\'evaluator\'],\n            \'output_root\':str(path_output),\'code_manifest_sha256\':pins[\'code\'][\'manifest_sha256\'],\n            \'source_transport_mode\':transport_mode,\'source_archive_bytes_verified\':transport_mode==\'verified-archive\',\n            \'entry_provenance\':entry_provenance}\n'
GUARD_SOURCE = '"""Linux lifecycle proof for one trusted, exclusively owned subprocess tree.\n\nUse ``OwnedProcessGuard.spawn(..., evidence_dir=...)`` for complete ownership.\nA dedicated, model-free supervisor is a subreaper: double forks and setsid do\nnot leave its tree. The coordinator is also a process-local subreaper so that a\nforced supervisor shutdown can reap *identified* adopted descendants. Neither\nrole changes a host setting or signals/reaps unrelated processes. All signals\nand descendant waits use pidfds; /proc start times bind recorded incarnations.\n\nThis is lifecycle/dataflow isolation for trusted code, not an adversarial OS\nsandbox. In particular it does not claim to survive arbitrary external SIGKILL\nof every supervisor. Missing evidence fails closed. Import has no side effects.\n"""\nfrom __future__ import annotations\n\nimport ctypes\nimport json\nimport math\nimport os\nfrom pathlib import Path\nimport signal\nimport subprocess\nimport sys\nimport time\n\nFORMAT = \'irisx/owned-process-closure-v1\'\nIDENTITY_KEYS = {\'pid\', \'birth\', \'session\', \'state\'}\nMAX_RECORD_BYTES = 4 * 1024 * 1024\n\n\ndef _read_process(pid):\n    try:\n        raw = (Path(\'/proc\') / str(pid) / \'stat\').read_text()\n    except (FileNotFoundError, ProcessLookupError):\n        return None\n    fields = raw[raw.rfind(\')\') + 2:].split()\n    return {\'pid\': pid, \'birth\': int(fields[19]), \'session\': int(fields[3]),\n            \'state\': fields[0], \'parent\': int(fields[1])}\n\n\ndef _identity(row):\n    return {key: row[key] for key in (\'pid\', \'birth\', \'session\', \'state\')}\n\n\ndef _key(row):\n    return row[\'pid\'], row[\'birth\']\n\n\ndef _same(row, identity):\n    return row is not None and _key(row) == _key(identity)\n\n\ndef _require_linux():\n    if (sys.platform != \'linux\' or not hasattr(os, \'pidfd_open\')\n            or not hasattr(signal, \'pidfd_send_signal\') or not hasattr(os, \'P_PIDFD\')):\n        raise RuntimeError(\'Linux pidfd signaling and waiting required; no PID-only fallback\')\n\n\ndef _prctl(option, value):\n    libc = ctypes.CDLL(None, use_errno=True)\n    if libc.prctl(ctypes.c_int(option), ctypes.c_ulong(value), 0, 0, 0) != 0:\n        error = ctypes.get_errno()\n        raise OSError(error, os.strerror(error))\n\n\ndef enable_subreaper():\n    """Affects only this calling process; never reaps any child by itself."""\n    _require_linux()\n    _prctl(36, 1)  # PR_SET_CHILD_SUBREAPER\n\n\ndef arm_parent_death(expected_parent, death_signal=signal.SIGKILL):\n    """Arm in the newly launched process, then close the parent-exit race."""\n    _require_linux()\n    _prctl(1, int(death_signal))  # PR_SET_PDEATHSIG\n    if os.getppid() != expected_parent:\n        os.kill(os.getpid(), death_signal)  # Only the current process.\n        return False\n    return True\n\n\ndef _write_record(path, value):\n    path = Path(path)\n    raw = json.dumps(value, sort_keys=True, separators=(\',\', \':\'), allow_nan=False).encode()\n    if len(raw) > MAX_RECORD_BYTES:\n        raise ValueError(\'owned-process record exceeds bound\')\n    temporary = path.with_name(path.name + \'.pending\')\n    with temporary.open(\'xb\') as stream:\n        stream.write(raw); stream.flush(); os.fsync(stream.fileno())\n    if path.exists():\n        raise ValueError(\'owned-process evidence identity consumed\')\n    os.replace(temporary, path)\n\n\ndef _read_record(path):\n    with Path(path).open(\'rb\') as stream:\n        raw = stream.read(MAX_RECORD_BYTES + 1)\n    if len(raw) > MAX_RECORD_BYTES:\n        raise ValueError(\'owned-process record exceeds bound\')\n    return json.loads(raw)\n\n\ndef _child_pids(pid):\n    """Parent edges from /proc stat; never inspect another process environment.\n\n    Some Linux containers omit task/*/children, so stat PPIDs are the portable\n    Linux source here. A discovered edge is rechecked before claiming ownership.\n    """\n    children = set()\n    for path in Path(\'/proc\').iterdir():\n        if not path.name.isdigit():\n            continue\n        row = _read_process(int(path.name))\n        if row is not None and row[\'parent\'] == pid:\n            children.add(row[\'pid\'])\n    return children\n\n\nclass OwnedProcessGuard:\n    """Track one Popen child; use spawn for a complete dedicated-tree proof.\n\n    A guard around an arbitrary preexisting Popen can stop identified processes,\n    but correctly reports ownership_complete=False: a preexisting orphan cannot\n    be attributed safely from a later process-table snapshot.\n    """\n    def __init__(self, child, *, _scope=None, _evidence_dir=None):\n        _require_linux()\n        self.child = child\n        row = _read_process(child.pid)\n        if row is None or row[\'parent\'] != os.getpid():\n            raise RuntimeError(\'guard requires an unreaped direct child\')\n        self.identity = _identity(row)\n        self._scope = _scope\n        self._evidence_dir = None if _evidence_dir is None else Path(_evidence_dir)\n        self._known = {}\n        self._pidfds = {}\n        self._reaped = set()\n        self._errors = []\n        self._forced = False\n        self._persisted = False\n        self._remember(row)\n        self.poll()\n\n    @classmethod\n    def spawn(cls, argv, *, evidence_dir, **kwargs):\n        """Launch the worker through a dedicated subreaper; preserve its argv/env.\n\n        stdout/stderr/cwd/env pass through to the worker. The supervisor itself\n        uses the caller\'s existing Python and only the standard library.\n        """\n        _require_linux()\n        if not isinstance(argv, (list, tuple)) or not argv or any(not isinstance(x, str) for x in argv):\n            raise ValueError(\'an explicit string argv is required\')\n        if kwargs.get(\'shell\') or kwargs.get(\'preexec_fn\') is not None or kwargs.get(\'pass_fds\'):\n            raise ValueError(\'shell, preexec_fn and inherited application fds are not supported\')\n        if kwargs.get(\'start_new_session\', True) is not True or kwargs.get(\'close_fds\', True) is not True:\n            raise ValueError(\'dedicated session and closed file descriptors required\')\n        evidence = Path(evidence_dir).absolute()\n        if any(p.is_symlink() for p in (evidence, *evidence.parents)):\n            raise ValueError(\'symlink process evidence path forbidden\')\n        evidence.mkdir(parents=True, exist_ok=False)\n        enable_subreaper()\n        kwargs[\'start_new_session\'] = True\n        kwargs[\'close_fds\'] = True\n        command = [sys.executable, \'-I\', \'-B\', \'-S\', str(Path(__file__).resolve()),\n                   \'--supervise\', str(os.getpid()), str(evidence), \'--\', *argv]\n        child = subprocess.Popen(command, **kwargs)\n        guard = cls(child, _evidence_dir=evidence)\n        try:\n            _write_record(evidence / \'launch.json\', {\n                \'format\': \'irisx/owned-process-launch-v1\', \'supervisor_identity\': guard.identity,\n                \'coordinator_identity\': _identity(_read_process(os.getpid())),\n                \'worker_argv\': list(argv), \'scope\': \'dedicated-subreaper-tree\'})\n            # Do not expose stop() until the supervisor\'s handlers and subreaper\n            # are armed. Otherwise an immediate TERM could bypass its cleanup.\n            deadline = time.monotonic() + 5\n            while not (evidence / \'worker-launch.json\').exists():\n                if guard.poll() is not None:\n                    raise RuntimeError(\'owned supervisor failed before launch handshake\')\n                if time.monotonic() >= deadline:\n                    raise TimeoutError(\'owned supervisor launch handshake timed out\')\n                time.sleep(0.005)\n        except BaseException:\n            guard.stop(timeout=5)\n            raise\n        return guard\n\n    def _remember(self, row):\n        key = _key(row)\n        if key in self._known:\n            return\n        self._known[key] = _identity(row)\n        try:\n            fd = os.pidfd_open(row[\'pid\'], 0)\n        except ProcessLookupError:\n            return\n        if not _same(_read_process(row[\'pid\']), row):\n            os.close(fd)\n            return\n        self._pidfds[key] = fd\n\n    def _discover(self):\n        roots = [self._known[key] for key in self._pidfds]\n        if self._scope is not None:\n            roots.append(self._scope)\n        visited = set()\n        while roots:\n            parent = roots.pop()\n            key = _key(parent)\n            if key in visited or not _same(_read_process(parent[\'pid\']), parent):\n                continue\n            visited.add(key)\n            for pid in _child_pids(parent[\'pid\']):\n                row = _read_process(pid)\n                # Confirm the parent edge while both incarnations still exist.\n                if (row is None or row[\'parent\'] != parent[\'pid\'] or row[\'birth\'] < parent[\'birth\']\n                        or not _same(_read_process(parent[\'pid\']), parent)):\n                    continue\n                self._remember(row)\n                roots.append(_identity(row))\n\n    def _reap(self):\n        self.child.poll()  # Popen alone owns its direct-child status.\n        for key in list(self._pidfds):\n            identity = self._known[key]\n            if identity[\'pid\'] == self.child.pid:\n                continue\n            row = _read_process(identity[\'pid\'])\n            fd = self._pidfds.get(key)\n            if (fd is None or not _same(row, identity) or row[\'parent\'] != os.getpid()\n                    or row[\'state\'] != \'Z\'):\n                continue\n            try:\n                result = os.waitid(os.P_PIDFD, fd, os.WEXITED | os.WNOHANG)\n                if result is not None:\n                    self._reaped.add(key)\n            except ChildProcessError:\n                pass  # Another owned ancestor, never wait for an arbitrary PID.\n        for key, fd in list(self._pidfds.items()):\n            if not _same(_read_process(key[0]), self._known[key]):\n                os.close(fd)\n                del self._pidfds[key]\n\n    def poll(self):\n        """Discover before and after reaping; return the supervisor/child status."""\n        self._discover()\n        self._reap()\n        self._discover()\n        return self.child.returncode\n\n    def _remaining(self):\n        return [_identity(row) for ident in self._known.values()\n                if _same((row := _read_process(ident[\'pid\'])), ident)]\n\n    def _scope_has_children(self):\n        if self._scope is None:\n            return False\n        # Only the exclusive supervisor makes this non-reaping presence probe.\n        # ECHILD, unlike a racing /proc snapshot, proves it has no children left.\n        try:\n            os.waitid(os.P_ALL, 0, os.WEXITED | os.WNOHANG | os.WNOWAIT)\n        except ChildProcessError:\n            return False\n        return True\n\n    def _signal(self, identity, signum):\n        key = _key(identity)\n        row = _read_process(identity[\'pid\'])\n        fd = self._pidfds.get(key)\n        if fd is None or not _same(row, identity) or row[\'state\'] == \'Z\':\n            return False\n        try:\n            signal.pidfd_send_signal(fd, signum, None, 0)\n            return True\n        except ProcessLookupError:\n            return False\n\n    def _stop_tree(self, deadline):\n        # Freeze known ancestors first. The exclusive subreaper retains any\n        # just-orphaned child, including new-session/double-fork descendants.\n        stable = None\n        while time.monotonic() < deadline:\n            self.poll()\n            remaining = self._remaining()\n            for identity in remaining:\n                self._signal(identity, signal.SIGSTOP)\n            self.poll()\n            after = self._remaining()\n            keys = frozenset(_key(row) for row in after)\n            if keys == stable and all(row[\'state\'] in (\'T\', \'t\', \'Z\') for row in after):\n                break\n            stable = keys\n            time.sleep(0.005)\n        # The root is killed last; pidfds prevent targeting a recycled PID.\n        self.poll()\n        rows = self._remaining()\n        rows.sort(key=lambda row: row[\'pid\'] == self.child.pid)\n        for identity in rows:\n            self._signal(identity, signal.SIGKILL)\n        while time.monotonic() < deadline:\n            self.poll()\n            # Adopted children may become discoverable/waitable just after their\n            # parent exits. Kill newly discovered owned incarnations as well.\n            remaining = self._remaining()\n            for identity in remaining:\n                self._signal(identity, signal.SIGKILL)\n            if self.child.returncode is not None and not remaining and not self._scope_has_children():\n                return\n            time.sleep(0.01)\n\n    def stop(self, timeout=10):\n        """Stop and reap this tree within timeout, returning a fail-closed proof."""\n        if not isinstance(timeout, (int, float)) or not math.isfinite(timeout) or timeout <= 0:\n            raise ValueError(\'positive finite stop timeout required\')\n        deadline = time.monotonic() + timeout\n        self._forced = True\n        self.poll()\n        if self._evidence_dir is not None and self.child.returncode is None:\n            # Let the dedicated supervisor retain/reap descendants itself.\n            self._signal(self.identity, signal.SIGTERM)\n            self._signal(self.identity, signal.SIGCONT)\n            grace = min(deadline, time.monotonic() + min(3.0, timeout / 2))\n            while time.monotonic() < grace and self.poll() is None:\n                time.sleep(0.01)\n        if self.child.returncode is None or self._remaining():\n            self._stop_tree(deadline)\n        return self.closed_proof()\n\n    def closed_proof(self):\n        """Snapshot proof; true closure requires all recorded incarnations gone."""\n        self.poll()\n        supervisor = None\n        errors = list(self._errors)\n        if self._evidence_dir is not None and self.child.returncode is not None:\n            try:\n                supervisor = _read_record(self._evidence_dir / \'supervisor-result.json\')\n                if (supervisor.get(\'format\') != \'irisx/owned-supervisor-result-v1\'\n                        or _key(supervisor[\'supervisor_identity\']) != _key(self.identity)):\n                    raise ValueError(\'foreign supervisor receipt\')\n                for identity in supervisor[\'owned_identities\']:\n                    if set(identity) != IDENTITY_KEYS:\n                        raise ValueError(\'invalid supervisor incarnation\')\n                    self._known.setdefault(_key(identity), identity)\n                errors.extend(supervisor[\'errors\'])\n            except (OSError, ValueError, KeyError, TypeError) as exc:\n                errors.append(\'supervisor receipt unavailable or invalid: \' + type(exc).__name__)\n        complete = (self._scope is not None or (supervisor is not None\n                    and supervisor.get(\'ownership_complete\') is True\n                    and supervisor.get(\'closed\') is True))\n        remaining = self._remaining()\n        proof = {\'format\': FORMAT, \'root_identity\': self.identity,\n                 \'scope\': \'dedicated-subreaper-tree\' if complete else \'observed-descendants-only\',\n                 \'ownership_complete\': complete,\n                 \'closed\': complete and self.child.returncode is not None and not remaining and not errors\n                           and not self._scope_has_children(),\n                 \'owned_identities\': sorted(self._known.values(), key=lambda row: (row[\'birth\'], row[\'pid\'])),\n                 \'remaining\': remaining, \'returncode\': self.child.returncode,\n                 \'worker_returncode\': (supervisor[\'worker_returncode\'] if supervisor is not None\n                                       else self.child.returncode if self._scope is not None else None),\n                 \'forced\': self._forced or bool(supervisor and supervisor[\'forced\']),\n                 \'reaped_identities\': [self._known[key] for key in sorted(self._reaped)],\n                 \'errors\': errors}\n        if self._evidence_dir is not None and proof[\'closed\'] and not self._persisted:\n            _write_record(self._evidence_dir / \'closure.json\', proof)\n            self._persisted = True\n        return proof\n\n\ndef verify_closed_proof(proof):\n    """Fresh incarnation checks for the seal/evaluator; raises on any ambiguity."""\n    required = {\'format\', \'root_identity\', \'scope\', \'ownership_complete\', \'closed\',\n                \'owned_identities\', \'remaining\', \'returncode\', \'worker_returncode\',\n                \'forced\', \'reaped_identities\', \'errors\'}\n    if (not isinstance(proof, dict) or set(proof) != required or proof[\'format\'] != FORMAT\n            or proof[\'scope\'] != \'dedicated-subreaper-tree\' or proof[\'closed\'] is not True\n            or proof[\'ownership_complete\'] is not True or proof[\'remaining\'] or proof[\'errors\']\n            or type(proof[\'returncode\']) is not int or type(proof[\'worker_returncode\']) is not int\n            or type(proof[\'forced\']) is not bool):\n        raise ValueError(\'complete owned-process closure proof required\')\n    root = proof[\'root_identity\']\n    identities = proof[\'owned_identities\']\n    if (not isinstance(root, dict) or set(root) != IDENTITY_KEYS\n            or not isinstance(identities, list) or not identities\n            or not any(row == root for row in identities)):\n        raise ValueError(\'bound owner incarnation required\')\n    keys = set()\n    for identity in identities:\n        if (not isinstance(identity, dict) or set(identity) != IDENTITY_KEYS\n                or type(identity[\'pid\']) is not int or identity[\'pid\'] <= 0\n                or type(identity[\'birth\']) is not int or identity[\'birth\'] < 0\n                or type(identity[\'session\']) is not int or type(identity[\'state\']) is not str\n                or _key(identity) in keys):\n            raise ValueError(\'invalid or duplicate owned incarnation\')\n        keys.add(_key(identity))\n        if _same(_read_process(identity[\'pid\']), identity):\n            raise ValueError(\'an owned process incarnation is still present\')\n    return True\n\n\ndef _supervise(expected_parent, evidence, argv):\n    stopped = [False]\n    def request_stop(_signum, _frame):\n        stopped[0] = True\n    signal.signal(signal.SIGTERM, request_stop)\n    signal.signal(signal.SIGINT, request_stop)\n    enable_subreaper()\n    if not arm_parent_death(expected_parent, signal.SIGTERM) or stopped[0]:\n        return 125\n    supervisor_identity = _identity(_read_process(os.getpid()))\n    command = [sys.executable, \'-I\', \'-B\', \'-S\', str(Path(__file__).resolve()),\n               \'--exec-worker\', str(os.getpid()), \'--\', *argv]\n    child = subprocess.Popen(command, close_fds=True)\n    guard = OwnedProcessGuard(child, _scope=supervisor_identity)\n    try:\n        _write_record(Path(evidence) / \'worker-launch.json\', {\n            \'format\': \'irisx/owned-worker-launch-v1\', \'worker_identity\': guard.identity,\n            \'supervisor_identity\': supervisor_identity, \'parent_death_signal\': int(signal.SIGKILL)})\n        while guard.poll() is None and not stopped[0]:\n            time.sleep(0.02)\n        proof = guard.closed_proof()\n        if stopped[0] or not proof[\'closed\']:\n            proof = guard.stop(timeout=5)\n    except BaseException:\n        proof = guard.stop(timeout=5)\n        proof[\'errors\'].append(\'supervisor exception\')\n        proof[\'closed\'] = False\n    result = {**proof, \'format\': \'irisx/owned-supervisor-result-v1\',\n              \'supervisor_identity\': supervisor_identity,\n              \'worker_returncode\': child.returncode}\n    _write_record(Path(evidence) / \'supervisor-result.json\', result)\n    if not proof[\'closed\'] or proof[\'errors\']:\n        return 125\n    if stopped[0] or proof[\'forced\']:\n        return 125 if child.returncode == 0 else (128 - child.returncode if child.returncode < 0 else child.returncode)\n    return 128 - child.returncode if child.returncode < 0 else child.returncode\n\n\ndef _main(argv):\n    if len(argv) >= 5 and argv[0] == \'--supervise\' and argv[3] == \'--\':\n        return _supervise(int(argv[1]), argv[2], argv[4:])\n    if len(argv) >= 4 and argv[0] == \'--exec-worker\' and argv[2] == \'--\':\n        if not arm_parent_death(int(argv[1])):\n            return 125\n        os.execvp(argv[3], argv[3:])\n    raise ValueError(\'invalid owned-process entrypoint\')\n\n\nif __name__ == \'__main__\':\n    raise SystemExit(_main(sys.argv[1:]))\n'
LOADER_SHA256 = '9a871c22935d0d158ee31b683632e717358bd69f7d37258ad4dd09811c845887'
GUARD_SHA256 = '8bc66ac80499203bca9f583f0d5ea5ecc5d8a131a2aa267f661060fc69e0580c'
```
